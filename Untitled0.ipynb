```python
import pandas as pd
import json

data = []

with open('/content/kindle_reviews.json', 'r') as file:
    for line in file:
        try:
            data.append(json.loads(line))
        except:
            pass

df = pd.DataFrame(data)

print(df.head())
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
df['sentiment'] = df['overall'].apply(
    lambda x: 1 if x >= 4 else 0
)
```

```python
print(df['sentiment'].value_counts())
```

```python
import matplotlib.pyplot as plt

df['sentiment'].value_counts().plot(kind='bar')

plt.title("Sentiment Distribution")

plt.xlabel("Sentiment")
plt.ylabel("Count")

plt.show()
```

```python
import re

def clean_text(text):

    text = str(text)

    text = text.lower()

    text = re.sub(r'[^a-zA-Z\s]', '', text)

    return text
```

```python
df['clean_review'] = df['reviewText'].apply(clean_text)
```

```python
print(df[['reviewText', 'clean_review']].head())
```

```python
from nltk.tokenize import word_tokenize

df['tokens'] = df['clean_review'].apply(word_tokenize)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

df['tokens'] = df['tokens'].apply(
    lambda words: [word for word in words if word not in stop_words]
)

print(df[['clean_review', 'tokens']].head())
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

df['tokens_no_stopwords'] = df['tokens'].apply(
    lambda words: [
        word for word in words
        if word not in stop_words
    ]
)

print(df[['tokens', 'tokens_no_stopwords']].head())
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

df['stemmed'] = df['tokens_no_stopwords'].apply(
    lambda words: [stemmer.stem(word) for word in words]
)

print(df[['tokens_no_stopwords', 'stemmed']].head())
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

df['lemmatized'] = df['tokens_no_stopwords'].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)

print(df[['tokens_no_stopwords', 'lemmatized']].head())
```
